```python
import numpy as np
array = np.array([1,2,3,4,5])
print(type(array))

# * The element is storing in this data type as default 
print(array.dtype)
```

```python
# ? Creating different integer type using numpy
array = np.array([1,2,3,4,5], dtype = np.int8)
print(array.nbytes)
```
